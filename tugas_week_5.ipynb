```python
# Nama: Ghaby Pattama Sinaga
# NIM: 4232501030
#jawaban soal nomor 1

jumlah_buku = 7
harga_buku = 18500
diskon_persen = 10
uang_bayar = 150000

# 1. Total belanja, diskon, jumlah yang harus dibayar
total_belanja = jumlah_buku * harga_buku
diskon = total_belanja * diskon_persen / 100
total_bayar = total_belanja - diskon

# 2. Uang kembalian
kembalian = uang_bayar - total_bayar

# 3. Iuran per orang (bilangan bulat) dan sisanya
iuran = int(total_bayar) // 4
sisa = int(total_bayar) % 4

# 4. Apakah uang cukup?
cukup = uang_bayar >= total_bayar

print("Total belanja   : Rp", total_belanja)
print("Diskon          : Rp", diskon)
print("Harus dibayar   : Rp", total_bayar)
print("Kembalian       : Rp", kembalian)
print("Iuran per orang : Rp", iuran)
print("Sisa            : Rp", sisa)
print("Uang cukup?     :", cukup)
```

```python
# Nama: Ghaby Pattama Sinaga
# NIM: 4232501030
#jawaban soal nomor 2

jumlah_panel = 12
daya_panel = 450        # Wp
jam_efektif = 5         # jam/hari
efisiensi = 0.8
kebutuhan = 25          # kWh/hari
tarif = 1444.70         # Rp/kWh

# 1. Energi per hari (kWh)
energi_harian = jumlah_panel * daya_panel * jam_efektif * efisiensi / 1000

# 2. Energi per bulan dan nilai rupiahnya
energi_bulanan = energi_harian * 30
nilai_rupiah = energi_bulanan * tarif

# 3. Kekurangan energi per hari
kekurangan = kebutuhan - energi_harian

# 4. Apakah PLTS mencukupi?
mencukupi = energi_harian >= kebutuhan

print("Energi per hari  :", energi_harian, "kWh")
print("Energi per bulan :", energi_bulanan, "kWh")
print("Nilai rupiah     : Rp", nilai_rupiah)
print("Kekurangan/hari  :", kekurangan, "kWh")
print("Mencukupi?       :", mencukupi)
```

```python
# Nama: Ghaby Pattama Sinaga
# NIM: 4232501030
#jawaban soal nomor 3

daya_generator = 200    # kW
jam_operasi = 8         # jam/hari
konsumsi_bbm = 0.25     # liter/kWh
harga_solar = 13500     # Rp/liter
kapasitas_tangki = 1000 # liter
arus = 20               # A
hambatan = 0.5          # ohm

# 1. Energi yang dibangkitkan per hari
energi = daya_generator * jam_operasi

# 2. Kebutuhan solar per hari dan biayanya
solar_per_hari = energi * konsumsi_bbm
biaya = solar_per_hari * harga_solar

# 3. Hari penuh tangki cukup dan sisa solar
hari_penuh = int(kapasitas_tangki // solar_per_hari)
sisa_solar = kapasitas_tangki % solar_per_hari

# 4. Rugi daya pada kabel
rugi_daya = arus ** 2 * hambatan

# 5. Rugi kabel tergolong tinggi?
rugi_tinggi = rugi_daya > 150 and arus >= 20

print("Energi per hari  :", energi, "kWh")
print("Solar per hari   :", solar_per_hari, "liter")
print("Biaya per hari   : Rp", biaya)
print("Tangki cukup     :", hari_penuh, "hari")
print("Sisa solar       :", sisa_solar, "liter")
print("Rugi daya kabel  :", rugi_daya, "W")
print("Rugi tinggi?     :", rugi_tinggi)
```
