```python
# Nama: Ghaby Pattama Sinaga
# NIM: 4232501030
#jawaban latihan

# 1 Program Menentukan Tahun Kabisat

def cek_kabisat(tahun):
    if (tahun % 4 == 0 and tahun % 100 != 0) or (tahun % 400 == 0):
        print(f"{tahun} adalah tahun kabisat")
    else:
        print(f"{tahun} bukan tahun kabisat")

# Daftar tahun yang diuji
daftar_tahun = [2024, 2023, 1900, 2000, 2026]

for tahun in daftar_tahun:
    cek_kabisat(tahun)

# 2 Program Kalkulator Belanja

# Lingkup global: bisa diakses oleh seluruh fungsi
PAJAK = 0.11  # PPN 11%

# Fungsi input (data sudah ditentukan)
def input_data():
    nama = "Buku Tulis"
    harga = 5000
    jumlah = 12
    print("===== DATA INPUT =====")
    print(f"Nama barang  : {nama}")
    print(f"Harga satuan : Rp {harga:,.0f}")
    print(f"Jumlah beli  : {jumlah}")
    return nama, harga, jumlah

# Fungsi proses (operasi aritmatika)
def hitung_total(harga, jumlah):
    # variabel lokal: hanya ada di dalam fungsi ini
    subtotal = harga * jumlah      # perkalian
    pajak = subtotal * PAJAK       # perkalian dengan variabel global
    total = subtotal + pajak       # penjumlahan
    return subtotal, pajak, total

# Fungsi output
def tampilkan_hasil(nama, jumlah, subtotal, pajak, total):
    print("\n===== STRUK BELANJA =====")
    print(f"Barang   : {nama} x {jumlah}")
    print(f"Subtotal : Rp {subtotal:,.0f}")
    print(f"Pajak    : Rp {pajak:,.0f}")
    print(f"Total    : Rp {total:,.0f}")

# Program utama
nama, harga, jumlah = input_data()
subtotal, pajak, total = hitung_total(harga, jumlah)
tampilkan_hasil(nama, jumlah, subtotal, pajak, total)
```
