```python
# Nama: Ghaby Pattama Sinaga
# NIM: 4232501030

# Latihan 1: Mengecek apakah angka merupakan kelipatan 5
angka = int(input("Masukkan sebuah angka: "))

if angka % 5 == 0:
    print("Kelipatan 5")
else:
    print(f"{angka} bukan kelipatan 5")
```

```python
# Latihan 2: Menentukan tahun kabisat menggunakan kondisi
# Syarat kabisat:
# - Habis dibagi 400 -> kabisat
# - Habis dibagi 4 dan tidak habis dibagi 100 -> kabisat

tahun = int(input("Masukkan tahun: "))

if tahun % 400 == 0:
    print(f"{tahun} adalah tahun kabisat")
elif tahun % 4 == 0 and tahun % 100 != 0:
    print(f"{tahun} adalah tahun kabisat")
else:
    print(f"{tahun} bukan tahun kabisat")
```
