```python
#aktivitas 6f
skor = [80, 65, 90, 75, 70]

# Tebakan saya:
# total akhir       -> 380
# rata-rata         -> 76.0
# skor >= 75        -> 3 dari 5

total = 0                    # akumulator: mulai dari nol
jumlah_lulus = 0             # penghitung = akumulator yang menambah 1
for nilai in skor:
    total = total + nilai    # tambahkan setiap elemen
    if nilai >= 75:
        jumlah_lulus = jumlah_lulus + 1
    print("total sementara:", total)

rata_rata = total / len(skor)
print("Total     :", total)
print("Rata-rata :", rata_rata)
print("Skor ≥ 75:", jumlah_lulus, "dari", len(skor))
```

```python
#aktivitas 6g
suhu = [28, 31, 30, 33, 29, 27, 32]

# Perubahan harian: mulai dari indeks 1 karena kita melihat i - 1
for i in range(1, len(suhu)):
    perubahan = suhu[i] - suhu[i - 1]
    print(f"hari {i + 1}: {suhu[i]} °C, perubahan {perubahan:+d}")

# Rata-rata bergerak 5 hari: jendela = 5 nilai yang berakhir di indeks i
rata_5 = []
for i in range(4, len(suhu)):          # mulai di indeks 4
    jendela = suhu[i - 4 : i + 1]      # 5 nilai: indeks i-4 sampai i
    rata_5.append(round(sum(jendela) / 5, 1))   # dibagi 5

print("Rata-rata 5 hari:", rata_5)
print(len(suhu), "nilai →", len(rata_5), "rata-rata")
```

```python
#aktivitas 6h
energi = [
    5576, 6098, 6078, 6095, 6074, 6077, 5564, 5552, 6100, 6083,
    6097, 6089, 6066, 5533, 5566, 6097, 3592, 6059, 6100, 6098,
    5565, 5571, 6103, 6135, 6065, 6071, 6086, 5577, 5575, 6062,
]

hari = 17
i = hari - 1                     # indeks hari ke-17

# 1) Besar penurunan dibanding hari sebelumnya
perubahan = (energi[i] - energi[i - 1]) / energi[i - 1] * 100
print(f"Hari {hari}: {energi[i]:,} MWh, hari sebelumnya {energi[i-1]:,} MWh")
print(f"Perubahan: {perubahan:+.1f}%")

# 2) Apakah hari ke-17 hari kerja atau akhir pekan?
#    Akhir pekan = hari 7-8, 14-15, 21-22, 28-29 (pola energi rendah)
akhir_pekan = [7, 8, 14, 15, 21, 22, 28, 29]
print("Hari ke-17 akhir pekan?", hari in akhir_pekan)

# 3) Bandingkan dengan rata-rata hari kerja dan akhir pekan
hari_kerja = [energi[h - 1] for h in range(1, 31)
              if h not in akhir_pekan and h != hari]
data_akhir_pekan = [energi[h - 1] for h in akhir_pekan]

rata_kerja = sum(hari_kerja) / len(hari_kerja)
rata_akhir = sum(data_akhir_pekan) / len(data_akhir_pekan)
print(f"Rata-rata hari kerja (tanpa hari 17): {rata_kerja:,.0f} MWh")
print(f"Rata-rata akhir pekan               : {rata_akhir:,.0f} MWh")
print(f"Hari 17 vs rata-rata hari kerja : {(energi[i] - rata_kerja) / rata_kerja * 100:+.1f}%")
print(f"Hari 17 vs rata-rata akhir pekan: {(energi[i] - rata_akhir) / rata_akhir * 100:+.1f}%")

# 4) (Opsional) Jika kamu punya list jam_operasi dari bagian data 30 hari,
#    salin list itu ke sel ini lalu hapus tanda # di dua baris berikut:
# print("jam_operasi hari 17:", jam_operasi[16])
# print("jam_operasi rata-rata:", sum(jam_operasi) / len(jam_operasi))
```

```python
#aktivitas 6i
# Data: 7 suhu harian (°C)
suhu = [29, 32, 31, 34, 30, 28, 33]
print("Data awal:", suhu)

# (1) Slicing: 3 hari pertama dan 2 hari terakhir
print("\n(1) 3 hari pertama :", suhu[:3])
print("    2 hari terakhir:", suhu[-2:])

# (2) Tambahkan satu hari dengan append, lalu keluarkan lagi dengan pop
suhu.append(30)
print("\n(2) Setelah append(30):", suhu)
dikeluarkan = suhu.pop()
print("    Hasil pop()       :", dikeluarkan)
print("    Setelah pop       :", suhu)

# (3) Versi urut dengan sorted, tanpa mengubah suhu
urut = sorted(suhu)
print("\n(3) Versi urut :", urut)
print("    suhu asli  :", suhu)

# (4) Rata-rata dan suhu tertinggi beserta hari ke berapa
total = 0
i_maks = 0
for i in range(len(suhu)):
    total = total + suhu[i]
    if suhu[i] > suhu[i_maks]:
        i_maks = i          # simpan POSISI, bukan hanya nilai
rata_rata = total / len(suhu)
print("\n(4) Rata-rata :", rata_rata, "°C")
print(f"    Tertinggi : {suhu[i_maks]} °C pada hari ke-{i_maks + 1}")

# (5) List baru berisi suhu di atas rata-rata
di_atas = []
for nilai in suhu:
    if nilai > rata_rata:
        di_atas.append(nilai)
print("\n(5) Suhu di atas rata-rata:", di_atas)
```

```python
#pemanasan
skor = [80, 65, 90, 75, 70, 85]

# (1) Skor pertama dan terakhir (indeks negatif untuk yang terakhir)
print("(1) Skor pertama :", skor[0])
print("    Skor terakhir:", skor[-1])

# (2) Rata-rata
total_skor = 0
for nilai in skor:
    total_skor = total_skor + nilai
rata_skor = total_skor / len(skor)
print("(2) Rata-rata    :", rata_skor)

# (3) Berapa skor di atas rata-rata
di_atas = 0
for nilai in skor:
    if nilai > rata_skor:
        di_atas = di_atas + 1
print("(3) Skor di atas rata-rata:", di_atas, "dari", len(skor))

# (4) Posisi skor terendah sebagai nomor mahasiswa (mahasiswa pertama = 1)
i_min = 0
for i in range(len(skor)):
    if skor[i] < skor[i_min]:
        i_min = i
print("(4) Skor terendah:", skor[i_min], "-> mahasiswa ke-", i_min + 1)

# (5) List baru dengan setiap skor ditambah 5
skor_baru = []
for nilai in skor:
    skor_baru.append(nilai + 5)
print("(5) Skor + 5     :", skor_baru)
```

```python
#latihan praktikum
energi = [
    5576, 6098, 6078, 6095, 6074, 6077, 5564, 5552, 6100, 6083,   # hari 1-10
    6097, 6089, 6066, 5533, 5566, 6097, 3592, 6059, 6100, 6098,   # hari 11-20
    5565, 5571, 6103, 6135, 6065, 6071, 6086, 5577, 5575, 6062,   # hari 21-30
]

print("=" * 50)
print("LAPORAN PRODUKSI ENERGI BULANAN")
print("=" * 50)

# 1) Total dan rata-rata energi
total = 0
for nilai in energi:
    total = total + nilai
rata = total / len(energi)
print(f"1) Total energi : {total:,} MWh")
print(f"   Rata-rata    : {rata:,.1f} MWh/hari")

# 2) Tiga hari tertinggi dan tiga hari terendah (nomor hari dan nilainya)
pasangan = []                                  # (nilai, nomor hari)
for i in range(len(energi)):
    pasangan.append((energi[i], i + 1))

tiga_rendah = sorted(pasangan)[:3]
tiga_tinggi = sorted(pasangan, reverse=True)[:3]

print("2) Tiga hari tertinggi:")
for nilai, hari in tiga_tinggi:
    print(f"   hari ke-{hari}: {nilai:,} MWh")
print("   Tiga hari terendah:")
for nilai, hari in tiga_rendah:
    print(f"   hari ke-{hari}: {nilai:,} MWh")

# 3) Jumlah hari di bawah target (target dimasukkan pengguna)
target = int(input("3) Masukkan target energi harian (MWh), misalnya 5800: "))
hari_kurang = []
for i in range(len(energi)):
    if energi[i] < target:
        hari_kurang.append(i + 1)
print(f"   Hari di bawah target {target:,} MWh: {len(hari_kurang)} hari")
print("   Nomor hari:", hari_kurang)

# 4) Hari yang rata-rata bergerak 7 harinya di bawah 97% rata-rata bulanan
batas = 0.97 * rata
rata_7 = []
for i in range(6, len(energi)):
    jendela = energi[i - 6 : i + 1]
    rata_7.append(sum(jendela) / 7)

print(f"4) Batas 97% rata-rata bulanan: {batas:,.1f} MWh")
print(f"   Jumlah rata-rata bergerak 7 hari: {len(rata_7)}")
ada = False
for k in range(len(rata_7)):
    hari = k + 7                               # jendela berakhir di hari ke-(k+7)
    if rata_7[k] < batas:
        print(f"   hari ke-{hari}: rata-rata 7 hari = {rata_7[k]:,.1f} MWh (di bawah batas)")
        ada = True
if not ada:
    print("   Tidak ada hari yang turun di bawah batas.")

# 5) Total energi per minggu (minggu 1-4, ditambah sisa hari)
print("5) Total energi per minggu:")
for m in range(4):
    awal = m * 7
    akhir = awal + 7
    print(f"   Minggu {m + 1} (hari {awal + 1}-{akhir}): {sum(energi[awal:akhir]):,} MWh")
sisa = energi[28:]
print(f"   Sisa hari (hari 29-30): {sum(sisa):,} MWh")
```
