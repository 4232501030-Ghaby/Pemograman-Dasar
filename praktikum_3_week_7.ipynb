```python
# === PRAKTIKUM 3A: Total Belanja ===
harga_barang = [15000, 25000, 10000, 5000, 8000,
                12000, 30000, 7500, 22000, 18000,
                9500, 4000]   # minimal 10 item
total_belanja = 1000

for harga in harga_barang:
    total_belanja += harga

print(f"Jumlah item  : {len(harga_barang)}")
print(f"Total belanja: Rp {total_belanja:,.0f}")
```

```python
# === PRAKTIKUM 3B: Iterasi String ===
nama_institusi = "Politeknik Negeri Batam"
hasil_output = ""

for karakter in nama_institusi:
    hasil_output += karakter + "-"

print(hasil_output)
# Output: P-o-l-i-t-e-k-n-i-k- -N-e-g-e-r-i- -B-a-t-a-m-
```

```python
# === PRAKTIKUM 3C: enumerate() dengan nama bulan ===
bulan = ["Januari", "Februari", "Maret", "April",
         "Mei", "Juni", "Juli", "Agustus",
         "September", "Oktober", "November", "Desember"]

for i, nama in enumerate(bulan):
    print(f"Bulan ke-{i + 1:>2}: {nama}")
```

```python
# === PRAKTIKUM 3D: Kelurahan di Kota Batam ===
kelurahan_batam = [
    "Belian", "Baloi Permai", "Taman Baloi", "Sukajadi",
    "Batu Selicin", "Lubuk Baja Kota", "Kampung Pelita",
    "Batu Merah", "Tanjung Uma", "Sungai Jodoh",
    "Nongsa", "Sambau"
]

print("Daftar Kelurahan di Kota Batam:")
print("-" * 35)
for i, nama in enumerate(kelurahan_batam):
    print(f"{i + 1:>3}. {nama}")
print(f"\nTotal: {len(kelurahan_batam)} kelurahan ditampilkan")
```
