```python
#soal nomor 2

daya_kw = 450
jam = 8
energi_aktual = 7200
kapasitas_kw = 400

# a. Energi harian
energi = daya_kw * jam
print(f"Energi harian = {energi} kWh")

# b. Capacity factor
capacity_factor = (energi_aktual / (kapasitas_kw * 24)) * 100
print(f"Capacity factor = {capacity_factor:.2f}%")
```

3. Capacity factor menunjukkan perbandingan hasil nyata dengan hasil maksimal. Hasil maksimal didapat dari `kapasitas_kw * 24`, yaitu 400 x 24 = 9600 kWh (kalau pembangkit bekerja penuh 24 jam). Lalu 7200 dibagi 9600 = 0,75, dikali 100 jadi 75%. Artinya pembangkit hanya dimanfaatkan 75% dari kemampuan maksimalnya.
4. `print(f"...")` menampilkan hasil ke layar. Huruf `f` di depan membuat nilai variabel di dalam kurung kurawal `{}` otomatis dimasukkan ke dalam teks.
5. `:.2f` artinya angka ditampilkan dengan 2 angka di belakang koma, sesuai permintaan soal.


Energi harian (3600 kWh) dan energi aktual (7200 kWh) angkanya berbeda karena keduanya memang data yang berbeda dari soal. Energi harian dipakai untuk bagian a, sedangkan energi aktual dipakai untuk menghitung capacity factor di bagian b.

```python
# soal nomor 6

suhu = float(input("Masukkan suhu uap (°C): "))
tekanan = float(input("Masukkan tekanan (bar): "))

if suhu >= 550 or tekanan >= 14:
    print("STATUS: TRIP")
elif suhu > 520 and tekanan > 12:
    print("STATUS: WARNING")
else:
    print("STATUS: NORMAL")
```

### Penjelasan Soal 6

**Cara kerja program:**
Program membaca suhu uap dan tekanan dari pengguna, lalu menentukan status boiler (TRIP, WARNING, atau NORMAL) menggunakan percabangan if-elif-else. Urutan pengecekan penting: kondisi paling berbahaya (TRIP) dicek lebih dulu, lalu WARNING, dan sisanya NORMAL.

**Fungsi setiap bagian kode:**
- `input()`: membaca masukan dari pengguna dalam bentuk teks.
- `float()`: mengubah teks menjadi angka desimal agar bisa dibandingkan.
- `if suhu >= 550 or tekanan >= 14`: operator `or` membuat kondisi terpenuhi jika salah satu benar, hasilnya STATUS: TRIP.
- `elif suhu > 520 and tekanan > 12`: operator `and` mengharuskan kedua kondisi benar, hasilnya STATUS: WARNING.
- `else`: dijalankan jika tidak ada kondisi di atas yang terpenuhi, hasilnya STATUS: NORMAL.

**Langkah-langkah:**
1. `input()` meminta pengguna mengetik angka. Hasil `input()` selalu berupa teks, jadi dibungkus `float()` untuk mengubahnya menjadi angka desimal supaya bisa dibandingkan.
2. Program memakai `if-elif-else`, yaitu percabangan yang mengecek kondisi dari atas ke bawah. Begitu ada satu kondisi yang benar, kondisi di bawahnya tidak dicek lagi.
3. Kondisi pertama: `suhu >= 550 or tekanan >= 14`. Kata `or` artinya cukup salah satu yang terpenuhi. Kalau suhu sudah 550 ke atas, atau tekanan sudah 14 ke atas, statusnya TRIP (kondisi paling bahaya).
4. Kondisi kedua: `suhu > 520 and tekanan > 12`. Kata `and` artinya dua-duanya harus terpenuhi. Kalau suhu di atas 520 dan tekanan di atas 12, statusnya WARNING (waspada).
5. `else` menangkap semua sisanya, yaitu STATUS: NORMAL.

**Alasan urutan penting:**
Contoh suhu 540 dan tekanan 14. Kondisi WARNING juga terpenuhi, tetapi karena tekanan sudah mencapai 14, statusnya harus TRIP. Karena itu TRIP dicek lebih dulu.


```python
# soal nomor 8

volume = 100

while volume > 0:
    volume -= 20
    print(f"Volume: {volume} liter")
    if volume == 20:
        print("WARNING: Hampir kosong!")

print("TANGKI KOSONG - Pengosongan berhenti.")
```

### Penjelasan Soal 8

**Cara kerja program:**
Program mensimulasikan pengosongan tangki dari 100 liter. Pada setiap putaran, volume dikurangi 20 liter lalu volume saat itu dicetak. Perulangan berhenti ketika tangki kosong, lalu pesan penutup ditampilkan.

**Fungsi setiap bagian kode:**
- `volume = 100`: volume awal tangki.
- `while volume > 0`: perulangan yang terus berjalan selama volume masih di atas 0.
- `volume -= 20`: mengurangi volume 20 liter setiap putaran (sama dengan `volume = volume - 20`).
- `print(f"Volume: {volume} liter")`: mencetak volume setelah dikurangi.
- `if volume == 20`: mengecek apakah volume tepat 20 liter, jika ya maka dicetak "WARNING: Hampir kosong!".
- `print("TANGKI KOSONG ...")`: berada di luar loop, sehingga hanya tercetak sekali setelah loop selesai.

**Langkah-langkah:**
1. `volume = 100` adalah isi awal tangki.
2. `while volume > 0:` adalah perulangan. Artinya "selama volume masih lebih dari 0, ulangi terus isi di dalamnya". Perulangan berhenti sendiri ketika volume sudah 0.
3. `volume -= 20` artinya volume dikurangi 20. Ini sama saja dengan menulis `volume = volume - 20`.
4. `print(f"Volume: {volume} liter")` menampilkan isi tangki setelah dikurangi.
5. `if volume == 20:` mengecek apakah isi tangki tepat 20 liter. Kalau iya, program mencetak "WARNING: Hampir kosong!". Perhatikan tanda `==` dengan dua sama dengan, yang artinya "apakah sama dengan", berbeda dengan `=` yang artinya "isi dengan".
6. Pesan "TANGKI KOSONG" ditulis di luar perulangan (tidak menjorok ke dalam), jadi hanya tercetak sekali setelah semua putaran selesai.


```python
# soal nomor 10

data = ["62.5", "58.0", "81.3", "N/A", "66.0"]

jumlah_valid = 0

for n, nilai in enumerate(data, start=1):
    try:
        suhu = float(nilai)
    except ValueError:
        print(f"Titik {n}: data tidak valid - pemeriksaan dihentikan")
        break

    jumlah_valid += 1

    if suhu > 70.0:
        print(f"Warning: Titik {n} suhu bearing tinggi!")

print(f"Jumlah data valid: {jumlah_valid}")
```

### Penjelasan Soal 10

**Cara kerja program:**
Data suhu bearing berupa teks diperiksa satu per satu menggunakan perulangan for. Setiap data diubah ke angka dengan try-except. Jika gagal diubah (misalnya "N/A"), pemeriksaan dihentikan dengan break. Jika suhu melebihi 70.0, program menampilkan peringatan. Di akhir, jumlah data valid ditampilkan.

**Fungsi setiap bagian kode:**
- `data = [...]`: list berisi data suhu dalam bentuk teks, satu di antaranya "N/A" karena sensor rusak.
- `jumlah_valid = 0`: variabel penghitung data yang berhasil diperiksa.
- `for n, nilai in enumerate(data, start=1)`: `enumerate` memberi nomor titik mulai dari 1 sambil mengambil setiap data.
- `try: suhu = float(nilai)`: mencoba mengubah teks menjadi angka.
- `except ValueError`: dijalankan jika konversi gagal. Program mencetak pesan data tidak valid.
- `break`: menghentikan perulangan seketika, sehingga data setelah "N/A" tidak diperiksa.
- `jumlah_valid += 1`: menambah penghitung setiap data berhasil diubah.
- `if suhu > 70.0`: menampilkan peringatan jika suhu bearing terlalu tinggi.
- `print(f"Jumlah data valid: ...")`: berada di luar loop, sehingga tampil setelah pemeriksaan selesai.

**Langkah-langkah:**
1. `data = ["62.5", "58.0", "81.3", "N/A", "66.0"]` adalah list berisi data dalam bentuk teks (ada tanda kutip), jadi belum bisa dihitung sebagai angka.
2. `jumlah_valid = 0` adalah penghitung. Setiap data berhasil diperiksa, angkanya ditambah 1.
3. `for n, nilai in enumerate(data, start=1):` membaca data satu per satu dari awal. `enumerate` memberi nomor urut sekaligus, dan `start=1` membuat nomor titik dimulai dari 1 (bukan 0), sesuai soal.
4. `try:` artinya "coba jalankan ini". Di dalamnya `float(nilai)` mencoba mengubah teks menjadi angka. Untuk "62.5" berhasil, tapi untuk "N/A" gagal karena bukan angka.
5. `except ValueError:` dijalankan kalau percobaan tadi gagal. Program mencetak pesan "data tidak valid - pemeriksaan dihentikan". Berkat `try-except`, program tidak crash, hanya berpindah ke penanganan ini.
6. `break` langsung menghentikan perulangan. Jadi titik 5 (66.0) tidak ikut diperiksa karena pemeriksaan berhenti di titik 4.
7. `jumlah_valid += 1` hanya dijalankan kalau konversi berhasil, karena kalau gagal program sudah keluar lewat `break`.
8. `if suhu > 70.0:` mencetak peringatan jika suhu melebihi 70. Hanya titik 3 (81.3) yang kena.
9. Baris terakhir menampilkan jumlah data valid. Karena tidak menjorok ke dalam loop, baris ini jalan sekali setelah pemeriksaan selesai.

```python
# CHALLENGE: Pemantauan Beban Genset & Estimasi Bahan Bakar
# Masalah:
#   Operator pembangkit perlu memantau beban genset tiap jam,
#   mengklasifikasikan kondisi beban, dan memperkirakan apakah
#   bahan bakar di tangki cukup untuk seluruh periode operasi.
#
# Cara kerja:
#   1. Data beban (kW) dibaca sebagai teks dari sensor, ada yang rusak ("N/A").
#   2. Setiap data diubah ke float dengan try-except. Data rusak dilewati.
#   3. Beban diklasifikasikan dengan if-elif-else terhadap kapasitas genset.
#   4. Dihitung total energi, rata-rata, beban puncak, dan capacity factor.
#   5. Simulasi pemakaian bahan bakar dengan while sampai periode selesai
#      atau tangki habis.

# Data dan parameter
data_beban = ["210", "255", "300", "N/A", "380", "455",
              "520", "430", "350", "290", "240", "180"]  # kW per jam
jam_mulai = 6            # pemantauan mulai pukul 06:00
kapasitas_kw = 500       # kapasitas genset (kW)
konsumsi_bbm = 0.27      # liter per kWh
tangki_liter = 800       # isi awal tangki (liter)

#  Penyimpanan hasil
beban_valid = []         # list beban yang valid
jam_valid = []           # list jam untuk beban yang valid
hitung_status = {"RENDAH": 0, "OPTIMAL": 0, "TINGGI": 0, "OVERLOAD": 0}

print("=== PEMANTAUAN BEBAN GENSET ===")

# 1. Baca dan klasifikasi data
for i, nilai in enumerate(data_beban):
    jam = jam_mulai + i
    try:
        beban = float(nilai)
    except ValueError:
        print(f"{jam:02d}:00 | Data sensor rusak ({nilai}) -> dilewati")
        continue

    persen = beban / kapasitas_kw * 100

    if persen > 100:
        status = "OVERLOAD"
    elif persen >= 80:
        status = "TINGGI"
    elif persen >= 40:
        status = "OPTIMAL"
    else:
        status = "RENDAH"

    hitung_status[status] += 1
    beban_valid.append(beban)
    jam_valid.append(jam)
    print(f"{jam:02d}:00 | {beban:6.1f} kW | {persen:6.2f}% | {status}")

# 2. Ringkasan statistik
total_kwh = sum(beban_valid)             # tiap data mewakili 1 jam
rata_rata = total_kwh / len(beban_valid)
puncak = max(beban_valid)
jam_puncak = jam_valid[beban_valid.index(puncak)]
capacity_factor = (total_kwh / (kapasitas_kw * len(beban_valid))) * 100

print("\n=== RINGKASAN ===")
print(f"Data valid          : {len(beban_valid)} dari {len(data_beban)}")
print(f"Total energi        : {total_kwh:.2f} kWh")
print(f"Beban rata-rata     : {rata_rata:.2f} kW")
print(f"Beban puncak        : {puncak:.2f} kW pada pukul {jam_puncak:02d}:00")
print(f"Capacity factor     : {capacity_factor:.2f}%")
for status, jumlah in hitung_status.items():
    print(f"Jam status {status:<8}: {jumlah}")

# 3. Simulasi bahan bakar dengan while
print("\n=== SIMULASI BAHAN BAKAR ===")
sisa = tangki_liter
idx = 0

while idx < len(beban_valid) and sisa > 0:
    pakai = beban_valid[idx] * konsumsi_bbm
    sisa = max(sisa - pakai, 0)
    print(f"{jam_valid[idx]:02d}:00 | Pakai {pakai:6.2f} L | Sisa {sisa:7.2f} L")
    idx += 1

if sisa == 0 and idx < len(beban_valid):
    print(f"PERINGATAN: Tangki habis pada pukul {jam_valid[idx - 1]:02d}:00, "
          f"{len(beban_valid) - idx} jam data belum terlayani!")
elif sisa == 0:
    print("PERINGATAN: Tangki habis tepat di akhir periode!")
else:
    print(f"Bahan bakar cukup. Sisa akhir: {sisa:.2f} liter")
```

### Penjelasan Challenge

**Masalah yang diselesaikan:**
Operator pembangkit perlu memantau beban genset tiap jam, untuk mengetahui kondisi bebannya, dan memastikan bahan bakar di tangki cukup untuk seluruh periode operasi.

**Cara kerja program:**
1. Data beban dari sensor dibaca sebagai teks, ada yang rusak ("N/A").
2. Setiap data diubah ke angka dengan try-except. Data rusak dilewati.
3. Beban dibandingkan dengan kapasitas genset lalu diklasifikasikan: RENDAH, OPTIMAL, TINGGI, atau OVERLOAD.
4. Program menghitung total energi, rata-rata, beban puncak, dan capacity factor.
5. Simulasi pemakaian bahan bakar dijalankan dengan while sampai data habis atau tangki kosong.

**Fungsi setiap bagian kode:**
- `data_beban`, `kapasitas_kw`, `konsumsi_bbm`, `tangki_liter`: data input dan parameter genset.
- `beban_valid`, `jam_valid`: list untuk menyimpan data beban dan jam yang lolos pengecekan.
- `hitung_status`: dictionary untuk menghitung jumlah jam pada setiap status.
- `for ... enumerate`: membaca data satu per satu sambil menghitung jam.
- `try-except ValueError`: menangani data sensor rusak agar program tidak error.
- `continue`: melewati data rusak dan lanjut ke data berikutnya.
- `if-elif-else`: menentukan status beban berdasarkan persentase terhadap kapasitas.
- `sum()`, `max()`, `.index()`: menghitung total energi, beban puncak, dan jam terjadinya beban puncak.
- `while`: simulasi pemakaian bahan bakar jam demi jam selama data masih ada dan tangki belum kosong.
- `max(sisa - pakai, 0)`: mencegah sisa bahan bakar bernilai negatif.
- Blok `if-elif-else` terakhir: menampilkan peringatan jika tangki habis, atau menampilkan sisa bahan bakar jika cukup.
